```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

```python
tokenizer = AutoTokenizer.from_pretrained("google/gemma-2b-it")
```

```python
tokenizer("This is working", return_tensors='pt').to(device="cuda")
```

```python
model = AutoModelForCausalLM.from_pretrained(
    "google/gemma-2b-it",
    torch_dtype=torch.bfloat16
)
```

```python
model.to(device="cuda:0")
```

```python
output = tokenizer.decode(model.generate(**tokenizer(str("This is working" + "\n" + "Double Check"), return_tensors='pt').to(device="cuda"), max_length=500)[0])
```

```python
print(output)
```
